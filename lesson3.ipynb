```python
import pandas as pd
```

```python
df = pd.read_csv("sale.csv")

# Удаляем ошибочный столбец, если он есть
df = df.drop(columns=["renevue"], errors="ignore")
```

```python
# Рассчитываем выручку
df["revenue"] = df["price"] * df["quantity"]
```

```python
# Рассчитываем показатели
total_revenue = df["revenue"].sum()
total_products = len(df)
total_quantity = df["quantity"].sum()
best_product = df.loc[df["revenue"].idxmax(), "product"]
leaders_quantity = df.sort_values(by=["quantity","revenue"], ascending=[False, True]).head(3)

# Создаём таблицу итогов
summary = pd.DataFrame({
    "Показатель": [
        "Общая выручка",
        "Количество товарных позиций",
        "Продано единиц",
        "Лидер по выручке"
    ],
    "Значение": [
        total_revenue,
        total_products,
        total_quantity,
        best_product
    ]
})
```

```python
# Лидеры продаж — товары с наибольшей выручкой
leaders = df.sort_values(by="revenue", ascending=False).head(3)
```

```python
with pd.ExcelWriter("sales_full_report.xlsx") as writer:
    df.to_excel(writer, sheet_name="Все продажи", index=False)
    leaders.to_excel(writer, sheet_name="Лидеры", index=False)
    summary.to_excel(writer, sheet_name="Итоги", index=False)
    leaders_quantity.to_excel(writer, sheet_name="Лидеры по количеству",
    index=False
)

print("Отчёт успешно создан!")
print("Общая выручка:", total_revenue)
print("Лидер по выручке:", best_product)
print("Лидеры по количеству проданных единиц:")
print(leaders_quantity[["product", "quantity", "revenue"]])

    
```
